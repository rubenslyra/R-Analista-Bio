## 1. O que é o Python?

### 1.1 Uma Breve História

O Python é uma linguagem de programação de alto nível, interpretada e de propósito geral. Criada em 1991 por Guido van Rossum, tornou-se uma das linguagens mais populares do mundo devido à sua sintaxe clara, legibilidade e vasto ecossistema de bibliotecas.

### 1.2 Por que Python na Bioinformática Forense?

- **Biopython**: biblioteca madura para manipulação de sequências, alinhamentos e bancos de dados biológicos
- **Ecossistema científico**: NumPy, Pandas, SciPy, scikit-learn formam uma pilha completa de análise
- **Automação e pipelines**: scripts Python integram ferramentas de linha de comando e APIs
- **Machine Learning**: scikit-learn, TensorFlow e PyTorch para classificação de amostras forenses
- **Processamento de NGS**: pysam, HTSeq, Biopython para dados de sequenciamento
- **Gratuito e open-source**: acessível a qualquer laboratório

### 1.3 Aplicações Forenses do Python

| Aplicação | Bibliotecas Relevantes |
|-----------|----------------------|
| Análise de sequências STR/SNP | Biopython, pysam |
| Processamento de dados NGS | HTSeq, pysam, samtools |
| Machine learning forense | scikit-learn, pandas |
| Metabolômica e toxicologia | pyOpenMS, mzmine-python |
| Automação de pipelines | Snakemake, Nextflow (DSL) |

## 3. Tipos de Dados Fundamentais

### 3.1 Tipos Básicos

````python
# Inteiro
inteiro = 42
print(f'Inteiro: {inteiro}, tipo: {type(inteiro)}')

# Float (decimal)
decimal = 3.14159
print(f'Float: {decimal}, tipo: {type(decimal)}')

# String (texto)
texto = 'DNA forense'
print(f'String: {texto}, tipo: {type(texto)}')

# Booleano
logico = True
print(f'Booleano: {logico}, tipo: {type(logico)}')

# None (ausência de valor, similar ao NULL/NA do R)
vazio = None
print(f'None: {vazio}, tipo: {type(vazio)}')
````

````python
# Verificando e convertendo tipos
x = '123'
print(f'É string? {isinstance(x, str)}')  # True
print(f'É numérico? {isinstance(x, (int, float))}')  # False

# Conversão explícita
y = int(x)       # 123
z = str(y)       # '123'
w = float('3.14') # 3.14

print(f'Convertido para int: {y}')
print(f'Convertido para str: {z}')
print(f'Convertido para float: {w}')

# Cuidado com conversões impossíveis!
try:
    invalido = int('abc')
except ValueError as e:
    print(f'Erro na conversão: {e}')
````

### 3.2 Strings e Manipulação de Sequências

Em bioinformática, strings representam sequências de DNA/RNA/proteínas:

````python
# Strings como sequências biológicas
sequencia_dna = 'ATCGATCGATCGATCG'

# Operações básicas com strings
print(f'Sequência: {sequencia_dna}')
print(f'Comprimento: {len(sequencia_dna)} bases')
print(f'Maiúsculas: {sequencia_dna.upper()}')
print(f'Minúsculas: {sequencia_dna.lower()}')

# Contagem de bases
print(f'\nContagem de bases:')
for base in ['A', 'T', 'C', 'G']:
    contagem = sequencia_dna.count(base)
    print(f'  {base}: {contagem} ({contagem/len(sequencia_dna)*100:.1f}%)')

# Complemento reverso (essencial para primers!)
complemento = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
comp_reverso = ''.join(complemento[base] for base in reversed(sequencia_dna))
print(f'\nComplemento reverso: {comp_reverso}')

# Fatiamento (slicing) — indexação começa em 0!
print(f'\nFatiamento:')
print(f'  Primeiras 4 bases: {sequencia_dna[:4]}')
print(f'  Últimas 4 bases: {sequencia_dna[-4:]}')
print(f'  Bases 4 a 8: {sequencia_dna[4:8]}')
print(f'  Reversa: {sequencia_dna[::-1]}')
````

## 4. Estruturas de Dados

### 4.1 Listas

Listas são coleções ordenadas e mutáveis. Podem conter elementos de tipos diferentes.

````python
# Criando listas
bases_dna = ['A', 'T', 'C', 'G']
pesos_moleculares = [313.2, 304.2, 289.2, 329.2]  # g/mol
marcadores_str = ['D3S1358', 'TH01', 'D21S11', 'D18S51']

# Listas mistas (perfil genético)
perfil_amostra = [
    'CASO_2026_001',      # ID
    'sangue',             # tipo
    25.4,                 # concentração ng/µL
    True,                 # validado
    [15, 17],             # alelos D3S1358
    [7, 9.3]              # alelos TH01
]

print(f'Bases DNA: {bases_dna}')
print(f'Perfil amostra: {perfil_amostra}')

# Acessando elementos (indexação começa em 0!)
print(f'\nPrimeira base: {bases_dna[0]}')       # 'A'
print(f'Última base: {bases_dna[-1]}')           # 'G'
print(f'Duas primeiras: {bases_dna[:2]}')        # ['A', 'T']
print(f'ID da amostra: {perfil_amostra[0]}')     # 'CASO_2026_001'
print(f'Alelos D3S1358: {perfil_amostra[4]}')    # [15, 17]
````

````python
# Operações com listas
alelos = [15, 17, 7, 9.3, 28, 30, 14, 17]

# Funções úteis
print(f'Comprimento: {len(alelos)}')
print(f'Soma: {sum(alelos)}')
print(f'Média: {sum(alelos)/len(alelos):.2f}')
print(f'Mínimo: {min(alelos)}')
print(f'Máximo: {max(alelos)}')
print(f'Ordenado: {sorted(alelos)}')
print(f'Únicos: {set(alelos)}')  # conjunto remove duplicatas

# Modificando listas
novos_alelos = alelos.copy()
novos_alelos.append(22)        # Adicionar ao final
novos_alelos.insert(0, 10)     # Inserir na posição 0
novos_alelos.remove(9.3)       # Remover primeira ocorrência
removido = novos_alelos.pop()  # Remover e retornar último

print(f'\nLista modificada: {novos_alelos}')
print(f'Elemento removido: {removido}')

# List comprehension (muito usado em bioinformática!)
quadrados = [x**2 for x in range(1, 6)]
print(f'\nQuadrados: {quadrados}')

# Filtrar alelos acima de threshold
threshold = 15
alelos_altos = [a for a in alelos if a >= threshold]
print(f'Alelos >= {threshold}: {alelos_altos}')
````

### 4.2 Tuplas

Tuplas são listas **imutáveis**. Úteis para dados que não devem ser alterados (coordenadas genômicas, pares de alelos).

````python
# Tuplas para dados fixos
par_bases = ('A', 'T')  # Par de Watson-Crick
posicao_genomica = ('chr11', 2159913, 2159990)  # chr, start, end
genotipo_th01 = (7, 9.3)  # Alelos heterozigotos

print(f'Par de bases: {par_bases}')
print(f'Posição TH01: {posicao_genomica}')
print(f'Genótipo TH01: {genotipo_th01}')

# Desempacotamento (unpacking)
cromossomo, inicio, fim = posicao_genomica
print(f'\nCromossomo: {cromossomo}')
print(f'Início: {inicio:,} bp')
print(f'Fim: {fim:,} bp')
print(f'Tamanho do amplicon: {fim - inicio} bp')

# Tuplas são imutáveis
try:
    par_bases[0] = 'G'  # Isso vai falhar!
except TypeError as e:
    print(f'\nErro esperado: {e}')
````

### 4.3 Dicionários

Dicionários armazenam pares chave-valor. São essenciais para representar perfis genéticos, metadados de amostras e configurações.

````python
# Dicionário representando um perfil genético forense
perfil_genetico = {
    'id_amostra': 'CASO_2026_001',
    'data_coleta': '2026-10-20',
    'tipo_amostra': 'sangue_periferico',
    'dna_concentracao_ng_ul': 25.4,
    'ratio_260_280': 1.82,
    'marcadores': {
        'D3S1358': {'alelos': [15, 17], 'rfus': [2845, 2156]},
        'TH01': {'alelos': [7, 9.3], 'rfus': [3012, 2789]},
        'D21S11': {'alelos': [28, 30], 'rfus': [3890, 3456]},
        'D18S51': {'alelos': [14, 17], 'rfus': [2678, 2345]},
        'vWA': {'alelos': [14, 18], 'rfus': [3201, 2890]}
    },
    'analista': 'Dr. Silva',
    'validado': True
}

# Acessando valores
print(f"ID: {perfil_genetico['id_amostra']}")
print(f"Concentração: {perfil_genetico['dna_concentracao_ng_ul']} ng/µL")
print(f"Alelos TH01: {perfil_genetico['marcadores']['TH01']['alelos']}")
print(f"RFUs D3S1358: {perfil_genetico['marcadores']['D3S1358']['rfus']}")

# Método seguro de acesso (não gera erro se chave não existir)
nota = perfil_genetico.get('observacoes', 'Sem observações')
print(f"Observações: {nota}")
````

````python
# Iterando sobre dicionários
print('=== Perfil Genético ===')
for marcador, dados in perfil_genetico['marcadores'].items():
    alelos_str = '/'.join(str(a) for a in dados['alelos'])
    rfu_max = max(dados['rfus'])
    print(f'  {marcador}: {alelos_str} (RFU máx: {rfu_max})')

# Adicionar/modificar entradas
perfil_genetico['fga'] = {'alelos': [22, 24], 'rfus': [4102, 3567]}
perfil_genetico['validado'] = False

# Verificar existência de chave
print(f"\nFGA presente? {'fga' in perfil_genetico}")
print(f"SE33 presente? {'SE33' in perfil_genetico}")

# Chaves, valores e itens
print(f"\nChaves do perfil: {list(perfil_genetico.keys())[:5]}...")
print(f"Número de campos: {len(perfil_genetico)}")
````

### 4.4 Conjuntos (Sets)

Conjuntos armazenam elementos únicos e não ordenados. Úteis para identificar alelos únicos, comparar perfis e operações de interseção/união.

````python
# Conjuntos para análise de alelos
alelos_suspeito = {15, 17, 7, 9.3, 28, 30, 14, 18}
alelos_vitima = {14, 16, 6, 9, 29, 31, 13, 17}

print(f'Alelos suspeito: {alelos_suspeito}')
print(f'Alelos vítima: {alelos_vitima}')

# Operações de conjuntos (fundamentais em comparação de perfis!)
compartilhados = alelos_suspeito & alelos_vitima  # Interseção
todos_alelos = alelos_suspeito | alelos_vitima     # União
exclusivos_suspeito = alelos_suspeito - alelos_vitima  # Diferença
simetrica = alelos_suspeito ^ alelos_vitima         # Diferença simétrica

print(f'\nAlelos compartilhados: {compartilhados}')
print(f'Todos os alelos: {todos_alelos}')
print(f'Exclusivos do suspeito: {exclusivos_suspeito}')
print(f'Diferença simétrica: {simetrica}')

# Remover duplicatas de uma lista
lista_com_duplicatas = [15, 17, 15, 7, 7, 9.3, 28]
unicos = list(set(lista_com_duplicatas))
print(f'\nOriginais: {lista_com_duplicatas}')
print(f'Únicos: {unicos}')
````

## 5. Importando e Exportando Dados

### 5.1 Leitura de Arquivos CSV com Pandas

````python
import pandas as pd

# Ler arquivo CSV
# df = pd.read_csv('dados_forenses.csv')

# Criar DataFrame de exemplo (amostras forenses)
dados = {
    'id': ['AM001', 'AM002', 'AM003', 'AM004', 'AM005'],
    'tipo': ['sangue', 'saliva', 'tecido', 'sangue', 'osso'],
    'dna_concentracao_ng_ul': [25.4, 18.7, 42.1, 8.3, 3.2],
    'ratio_260_280': [1.82, 1.79, 1.85, 1.65, 1.45],
    'degradado': [False, False, False, True, True],
    'data_analise': pd.to_datetime(['2026-10-01', '2026-10-02', 
                                     '2026-10-03', '2026-10-04', '2026-10-05'])
}

df = pd.DataFrame(dados)

# Explorando o DataFrame
print('=== Primeiras linhas ===')
display(df.head())

print('\n=== Informações ===')
df.info()

print('\n=== Estatísticas descritivas ===')
display(df.describe())
````

````python
# Selecionando colunas
print('IDs e tipos:')
print(df[['id', 'tipo']])

# Filtrando linhas
print('\nAmostras de sangue:')
sangue = df[df['tipo'] == 'sangue']
display(sangue)

print('\nAmostras não degradadas com boa qualidade:')
boas = df[(df['degradado'] == False) & (df['ratio_260_280'] >= 1.8)]
display(boas)

# Adicionando nova coluna
df['qualidade'] = df['ratio_260_280'].apply(
    lambda x: 'Alta' if x >= 1.8 else ('Média' if x >= 1.6 else 'Baixa')
)
print('\nCom classificação de qualidade:')
display(df[['id', 'ratio_260_280', 'qualidade']])
````

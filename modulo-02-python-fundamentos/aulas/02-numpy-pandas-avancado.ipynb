## 1. NumPy: Computação Numérica Eficiente

### 1.1 Por que NumPy na Bioinformática?

- **Performance**: operações vetorizadas são 10-100x mais rápidas que loops Python
- **Memória**: arrays compactos e contíguos na memória
- **Álgebra linear**: essencial para análises multivariadas (PCA, PLS-DA)
- **Broadcasting**: operações entre arrays de dimensões diferentes
- **Integração**: base para Pandas, scikit-learn, SciPy

```python
import numpy as np
import pandas as pd

# Criando arrays NumPy
# Array 1D: frequências alélicas
freq_alelicas = np.array([0.23, 0.19, 0.08, 0.11, 0.32, 0.05, 0.02])
print(f'Frequências alélicas: {freq_alelicas}')
print(f'Tipo: {type(freq_alelicas)}')
print(f'Shape: {freq_alelicas.shape}')
print(f'Dtype: {freq_alelicas.dtype}')
```

### 1.2 Operações Vetorizadas vs Loops

A vetorização é fundamental para processar grandes volumes de dados forenses.

```python
import time

# Simular dataset grande: 100.000 amostras com RFUs
np.random.seed(42)
rfus_grande = np.random.normal(3000, 500, size=100000)

# Método 1: Loop Python (LENTO)
start = time.time()
resultado_loop = []
for rfu in rfus_grande:
    resultado_loop.append(rfu * 1.5 + 100)
tempo_loop = time.time() - start

# Método 2: Vetorizado NumPy (RÁPIDO)
start = time.time()
resultado_numpy = rfus_grande * 1.5 + 100
tempo_numpy = time.time() - start

print(f'Loop Python: {tempo_loop:.4f}s')
print(f'NumPy vetorizado: {tempo_numpy:.4f}s')
print(f'Aceleração: {tempo_loop/tempo_numpy:.1f}x')
print(f'Resultados idênticos? {np.allclose(resultado_loop, resultado_numpy)}')
```

### 1.3 Estatísticas Descritivas com NumPy

Cálculos essenciais para controle de qualidade em análises forenses.

```python
# Dados de concentração de DNA (ng/µL) de 50 amostras
concentracoes = np.random.uniform(0.5, 80, size=50)

print('=== Estatísticas Descritivas ===')
print(f'Média: {np.mean(concentracoes):.2f} ng/µL')
print(f'Mediana: {np.median(concentracoes):.2f} ng/µL')
print(f'Desvio padrão: {np.std(concentracoes):.2f}')
print(f'Variância: {np.var(concentracoes):.2f}')
print(f'Mínimo: {np.min(concentracoes):.2f}')
print(f'Máximo: {np.max(concentracoes):.2f}')
print(f'Amplitude: {np.ptp(concentracoes):.2f}')
print(f'Percentil 25: {np.percentile(concentracoes, 25):.2f}')
print(f'Percentil 75: {np.percentile(concentracoes, 75):.2f}')

# Contagem de amostras por faixa de concentração
baixa = np.sum(concentracoes < 5)
media = np.sum((concentracoes >= 5) & (concentracoes < 20))
alta = np.sum(concentracoes >= 20)

print(f'\n=== Classificação ===')
print(f'Baixa (<5 ng/µL): {baixa} amostras ({baixa/len(concentracoes)*100:.1f}%)')
print(f'Média (5-20 ng/µL): {media} amostras ({media/len(concentracoes)*100:.1f}%)')
print(f'Alta (≥20 ng/µL): {alta} amostras ({alta/len(concentracoes)*100:.1f}%)')
```

### 1.4 Álgebra Linear para Análises Forenses

Operações matriciais são usadas em cálculos de likelihood ratio, PCA e análise de misturas.

```python
# Matriz de distâncias genéticas entre populações
# Valores simulados de Fst (fixation index)
populacoes = ['Europeu', 'Africano', 'Asiático', 'Americano']
fst_matrix = np.array([
    [0.000, 0.120, 0.095, 0.045],  # Europeu
    [0.120, 0.000, 0.180, 0.135],  # Africano
    [0.095, 0.180, 0.000, 0.110],  # Asiático
    [0.045, 0.135, 0.110, 0.000]   # Americano
])

print('Matriz Fst entre populações:')
for i, pop in enumerate(populacoes):
    print(f'{pop}: {fst_matrix[i]}')

# Autovalores e autovetores (base do PCA)
autovalores, autovetores = np.linalg.eig(fst_matrix)

print(f'\nAutovalores (variância explicada): {np.round(autovalores.real, 4)}')
print(f'Proporção de variância: {np.round(autovalores.real / np.sum(autovalores.real), 4)}')

# Produto matricial: cálculo de scores de ancestralidade
pesos_ancestralidade = np.array([0.3, 0.4, 0.2, 0.1])  # pesos por marcador
scores = fst_matrix @ pesos_ancestralidade
print(f'\nScores de diferenciação: {np.round(scores, 4)}')
```

### 1.5 Broadcasting: Operações entre Arrays de Diferentes Dimensões

```python
# Normalizar matriz de RFUs por marcador
# Matriz: 5 amostras × 4 marcadores
rfu_matrix = np.array([
    [2845, 3012, 3890, 2678],
    [2567, 3201, 4102, 2345],
    [3102, 2789, 3456, 2890],
    [2234, 2890, 3567, 2567],
    [2678, 3012, 3890, 2345]
])

marcadores_rfu = ['D3S1358', 'TH01', 'D21S11', 'D18S51']

# Calcular média por marcador (eixo 0 = linhas)
medias_marcador = np.mean(rfu_matrix, axis=0)
print(f'Médias por marcador: {np.round(medias_marcador, 1)}')

# Broadcasting: subtrair média de cada coluna
rfu_normalizado = rfu_matrix - medias_marcador
print(f'\nRFU normalizado (desvio da média):\n{np.round(rfu_normalizado, 1)}')

# Normalização Z-score por marcador
dp_marcador = np.std(rfu_matrix, axis=0)
z_scores = (rfu_matrix - medias_marcador) / dp_marcador
print(f'\nZ-scores:\n{np.round(z_scores, 2)}')
```

## 2. Pandas Avançado

### 2.1 GroupBy: Agregações Complexas

Essencial para relatórios QC e análises estratificadas.

```python
# Criar dataset forense completo
np.random.seed(42)
n_amostras = 100

df_forense = pd.DataFrame({
    'id': [f'AM{i:03d}' for i in range(1, n_amostras + 1)],
    'tipo': np.random.choice(['sangue', 'saliva', 'tecido', 'osso'], n_amostras,
                             p=[0.35, 0.25, 0.25, 0.15]),
    'dna_conc': np.round(np.random.uniform(0.5, 80, n_amostras), 2),
    'ratio_260_280': np.round(np.random.uniform(1.2, 2.1, n_amostras), 2),
    'degradado': np.random.choice([True, False], n_amostras, p=[0.2, 0.8]),
    'analista': np.random.choice(['Silva', 'Santos', 'Oliveira', 'Costa'], n_amostras),
    'data_analise': pd.date_range('2026-09-01', periods=n_amostras, freq='D'),
    'prioridade': np.random.choice(['alta', 'media', 'baixa'], n_amostras,
                                   p=[0.2, 0.5, 0.3])
})

# Adicionar classificação de qualidade
df_forense['qualidade'] = pd.cut(
    df_forense['ratio_260_280'],
    bins=[0, 1.6, 1.8, 3.0],
    labels=['Baixa', 'Media', 'Alta']
)

print(f'Dataset criado: {df_forense.shape[0]} amostras × {df_forense.shape[1]} variáveis')
df_forense.head()
```

```python
# GroupBy simples: estatísticas por tipo de amostra
stats_por_tipo = df_forense.groupby('tipo')['dna_conc'].agg(
    n='count',
    media='mean',
    mediana='median',
    dp='std',
    min='min',
    max='max'
).round(2)

print('=== Estatísticas de Concentração por Tipo ===')
display(stats_por_tipo)
```

```python
# GroupBy múltiplo: tipo × qualidade
stats_cruzadas = df_forense.groupby(['tipo', 'qualidade']).agg(
    n_amostras=('id', 'count'),
    conc_media=('dna_conc', 'mean'),
    prop_degradadas=('degradado', 'mean')
).round(3)

print('=== Análise Cruzada: Tipo × Qualidade ===')
display(stats_cruzadas)
```

```python
# GroupBy com funções customizadas
def coeficiente_variacao(x):
    """CV = DP / Média × 100"""
    return (x.std() / x.mean()) * 100 if x.mean() != 0 else np.nan

stats_avancadas = df_forense.groupby('analista')['dna_conc'].agg(
    n='count',
    media='mean',
    cv_percentual=coeficiente_variacao
).round(2)

print('=== Performance por Analista ===')
display(stats_avancadas)
print('\nNota: CV baixo indica consistência nas extrações do analista')
```

### 2.2 Merge/Join: Combinando Datasets Forenses

Fundamental para integrar dados de diferentes fontes (perfis, metadados, resultados).

```python
# Dataset 1: Perfis genéticos
perfis = pd.DataFrame({
    'amostra_id': ['AM001', 'AM002', 'AM003', 'AM004', 'AM005'],
    'D3S1358': ['15/17', '14/18', '15/15', '16/18', '14/15'],
    'TH01': ['7/9.3', '6/9', '8/8', '7/7', '9/9.3'],
    'D21S11': ['28/30', '29/31', '28/28', '30/31', '28/29']
})

# Dataset 2: Metadados das amostras
metadados = pd.DataFrame({
    'amostra_id': ['AM001', 'AM002', 'AM003', 'AM005', 'AM006'],
    'tipo': ['sangue', 'saliva', 'tecido', 'sangue', 'osso'],
    'caso': ['CASO_001', 'CASO_001', 'CASO_002', 'CASO_003', 'CASO_003'],
    'data_coleta': ['2026-10-01', '2026-10-01', '2026-10-05', '2026-10-10', '2026-10-10']
})

print('=== Perfis Genéticos ===')
display(perfis)
print('\n=== Metadados ===')
display(metadados)
```

```python
# INNER JOIN: apenas amostras presentes em ambos
inner = pd.merge(perfis, metadados, on='amostra_id', how='inner')
print('=== INNER JOIN ===')
display(inner)
print(f'Resultado: {len(inner)} registros (apenas AM001-AM005 presentes em ambos)')
```

```python
# LEFT JOIN: todos os perfis + metadados quando disponível
left = pd.merge(perfis, metadados, on='amostra_id', how='left')
print('=== LEFT JOIN ===')
display(left)
print(f'\nNota: AM004 tem perfil mas não tem metadados (NaN)')
```

```python
# OUTER JOIN: todos os registros de ambos
outer = pd.merge(perfis, metadados, on='amostra_id', how='outer', indicator=True)
print('=== OUTER JOIN com indicador de origem ===')
display(outer)
print(f'\nContagem por origem:')
print(outer['_merge'].value_counts())
```

### 2.3 Pivot Tables: Relatórios Tabulares Forenses

Transformar dados longos em tabelas resumidas para laudos.

```python
# Pivot table: contagem de amostras por tipo e qualidade
pivot_contagem = pd.pivot_table(
    df_forense,
    values='id',
    index='tipo',
    columns='qualidade',
    aggfunc='count',
    fill_value=0,
    margins=True,
    margins_name='Total'
)

print('=== Contagem de Amostras: Tipo × Qualidade ===')
display(pivot_contagem)
```

```python
# Pivot table: concentração média por analista e prioridade
pivot_conc = pd.pivot_table(
    df_forense,
    values='dna_conc',
    index='analista',
    columns='prioridade',
    aggfunc=['mean', 'count'],
    round_=2
)

print('=== Concentração Média por Analista e Prioridade ===')
display(pivot_conc)
```

```python
# Pivot table com função customizada
pivot_cv = pd.pivot_table(
    df_forense,
    values='dna_conc',
    index='tipo',
    columns='analista',
    aggfunc=coeficiente_variacao
).round(1)

print('=== Coeficiente de Variação (%) por Tipo e Analista ===')
display(pivot_cv)
print('\nInterpretação: CV < 30% geralmente aceitável em análises forenses')
```

### 2.4 Transformações Wide ↔ Long no Pandas

Alternativa ao tidyr do R, essencial para perfis STR.

```python
# Formato WIDE: perfis STR típicos de laudos
perfis_wide = pd.DataFrame({
    'amostra': ['REF_001', 'QST_001', 'QST_002'],
    'D3S1358': ['15/17', '14/16', '15/15'],
    'TH01': ['7/9.3', '6/9', '8/9.3'],
    'D21S11': ['28/30', '29/31', '28/28'],
    'D18S51': ['14/17', '13/16', '15/17']
})

print('Formato WIDE:')
display(perfis_wide)
```

```python
# Wide → Long (melt)
perfis_long = pd.melt(
    perfis_wide,
    id_vars=['amostra'],
    var_name='marcador',
    value_name='genotipo'
)

print('Formato LONG:')
display(perfis_long)
```

```python
# Separar alelos em colunas distintas
perfis_long[['alelo1', 'alelo2']] = perfis_long['genotipo'].str.split('/', expand=True)

print('Com alelos separados:')
display(perfis_long)
```

```python
# Long → Wide (pivot)
perfis_volta_wide = perfis_long.pivot(
    index='amostra',
    columns='marcador',
    values='genotipo'
).reset_index()

print('De volta ao formato WIDE:')
display(perfis_volta_wide)

# Verificar integridade
identicos = perfis_wide.equals(perfis_volta_wide)
print(f'\nDados recuperados são idênticos? {identicos}')
```

## 3. Aplicações Forenses Práticas

### 3.1 Cálculo de Frequências Alélicas

```python
# Dataset de genótipos para cálculo de frequências
np.random.seed(42)
n_ind = 200

# Simular alelos TH01 (baseado em frequências reais brasileiras)
alelos_possiveis = [6, 7, 8, 9, 9.3, 10, 11]
freqs_reais = [0.23, 0.19, 0.08, 0.11, 0.32, 0.05, 0.02]

# Gerar genótipos diploides
alelo1 = np.random.choice(alelos_possiveis, n_ind, p=freqs_reais)
alelo2 = np.random.choice(alelos_possiveis, n_ind, p=freqs_reais)

df_genotipos = pd.DataFrame({
    'individuo': [f'IND_{i:03d}' for i in range(1, n_ind + 1)],
    'alelo1': alelo1,
    'alelo2': alelo2
})

# Calcular frequências alélicas observadas
todos_alelos = pd.concat([df_genotipos['alelo1'], df_genotipos['alelo2']])
freq_obs = todos_alelos.value_counts(normalize=True).sort_index()

print('=== Frequências Alélicas Observadas (TH01) ===')
for alelo, freq in freq_obs.items():
    print(f'  Alelo {alelo}: {freq:.4f}')

# Comparar com frequências esperadas
freq_df = pd.DataFrame({
    'alelo': alelos_possiveis,
    'freq_esperada': freqs_reais,
    'freq_observada': [freq_obs.get(a, 0) for a in alelos_possiveis]
})
freq_df['diferenca'] = freq_df['freq_observada'] - freq_df['freq_esperada']

print('\n=== Comparação Esperado vs Observado ===')
display(freq_df.round(4))
```

### 3.2 Heterozigosidade e Equilíbrio Hardy-Weinberg

```python
# Calcular heterozigosidade observada e esperada
heterozigotos = (df_genotipos['alelo1'] != df_genotipos['alelo2']).sum()
ho = heterozigotos / n_ind

# He = 1 - Σ(pi²)
he = 1 - sum(freqs_reais ** 2)

print(f'Heterozigosidade Observada (Ho): {ho:.4f}')
print(f'Heterozigosidade Esperada (He): {he:.4f}')
print(f'Fis (déficit de heterozigotos): {(he - ho) / he:.4f}')

# Teste exato de Hardy-Weinberg (simplificado)
# Em prática forense, usar scipy.stats ou pacote especializado
from scipy import stats

# Contagens genotípicas observadas
genotipos_obs = {}
for _, row in df_genotipos.iterrows():
    g = tuple(sorted([row['alelo1'], row['alelo2']]))
    genotipos_obs[g] = genotipos_obs.get(g, 0) + 1

print(f'\nGenótipos únicos observados: {len(genotipos_obs)}')
print(f'Homozigotos: {sum(1 for g in genotipos_obs if g[0] == g[1])}')
print(f'Heterozigotos: {sum(1 for g in genotipos_obs if g[0] != g[1])}')
```

### 3.3 Análise de Misturas de DNA

Identificação de contribuintes em misturas usando álgebra linear.

```python
# Simular mistura de 2 contribuintes
# RFUs observados na mistura
marcadores_mix = ['D3S1358', 'TH01', 'D21S11', 'D18S51']
rfu_mistura = np.array([4500, 3800, 5200, 4100])

# Perfis dos suspeitos (referência)
suspeito_a = np.array([2800, 2200, 3100, 2500])  # Contribuinte majoritário
suspeito_b = np.array([1700, 1600, 2100, 1600])  # Contribuinte minoritário

# Estimar proporção de mistura por mínimos quadrados
# rfu_mistura ≈ α × suspeito_a + β × suspeito_b
A = np.column_stack([suspeito_a, suspeito_b])
proporcoes, residuals, rank, s = np.linalg.lstsq(A, rfu_mistura, rcond=None)

alpha, beta = proporcoes
total = alpha + beta

print('=== Análise de Mistura ===')
print(f'Contribuição Suspeito A: {alpha/total*100:.1f}%')
print(f'Contribuição Suspeito B: {beta/total*100:.1f}%')
print(f'Ratio majoritário/minoritário: {alpha/beta:.2f}')

# Verificar ajuste
rfu_estimado = alpha * suspeito_a + beta * suspeito_b
erro_quadratico = np.sqrt(np.mean((rfu_mistura - rfu_estimado)**2))
print(f'\nRMSE do modelo: {erro_quadratico:.1f} RFU')

# Balanceamento alélico dentro da mistura
balance_ratio = min(alpha, beta) / max(alpha, beta)
print(f'Balance ratio: {balance_ratio:.2f}')
if balance_ratio >= 0.6:
    print('→ Mistura equilibrada (interpretação mais confiável)')
else:
    print('→ Mistura desequilibrada (atenção a drop-out alélico)')
```
